```python
# Cell 2: Decode ZATCA Phase 2 E-Invoice QR Code
sample_qr = "ARXYtNix2YPYqSDYp9mE2LPYrdin2KgCDzMxMDEyMzQ1Njc4MDAwMwMUMjAyNi0wMy0yNVQxMjowMDowMFoEBzExNTAuMDAFBjE1MC4wMA=="
!php bin/eidcloud-ar-ocr zatca-qr "$sample_qr" --decode

print("\n--- Outputting as JSON ---")
!php bin/eidcloud-ar-ocr zatca-qr "$sample_qr" --json
```
